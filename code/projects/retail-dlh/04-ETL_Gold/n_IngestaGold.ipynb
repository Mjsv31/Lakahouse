# Capa gold - Carga inicial

```python
import os
from spark_session import *
from dotenv import load_dotenv
```

## Variables

```python
load_dotenv()
vPathLand = os.getenv("PATH_LANDING")
vPolarisCatalog = os.getenv("POLARIS_CATALOG")
vSchemaGold = os.getenv("SCHEMA_GOLD")
```

## SparkSession

```python
spark = create_spark_session(
    "retail-dlh-carga-inicial-gold",
    configs=[
        configure_iceberg_extensions,
        configure_polaris_restcatalog,
        configure_retail_polaris,
        configure_s3
    ]
)
```

## Dimension de Tiempo

### Extracción

```python
# - Obtiene dataframe de la capa landing
df_DimDates=spark.read.format("parquet") \
    .load(f"{vPathLand}dim_dates.parquet")
```

### Carga

```python
# - Reemplaza la información en la capa gold
df_DimDates.write.format("iceberg")\
    .mode("overwrite")\
    .save(f"{vPolarisCatalog}.{vSchemaGold}.DIM_DATES")
```

## Cierre de SparkSession

```python
spark.stop()
```
