# Capa Bronze - Catálogo de Distribuidores

```python
from spark_session import *
```

## SparkSession

```python
spark = create_spark_session(
    "retail-dlh-bronze",
    configs=[
        configure_delta_extensions,
        configure_delta_catalog,
        configure_s3
    ]
)
```

## Extracción

```python
# - Obtiene dataframe de la capa landing
df_dealerships = spark.table("landing.dealerships")
```

## Carga

```python
# - Reemplaza la información en la capa bronze
df_dealerships.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze.dealership")
```
